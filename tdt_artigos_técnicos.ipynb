```python
import requests
from bs4 import BeautifulSoup

def extract_text_from_url(url):
  response = requests.get(url)

  if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    for script_or_style in soup(['script', 'style']):
      script_or_style.decompose()
    texto = soup.get_text(separator= '')
    #Limpar texto
    linhas = (line.strip() for line in texto.splitlines())
    parts = (phrase.strip() for line in linhas for phrase in line.split("  "))
    texto_limpo = '\n'.join(part for part in parts if part)
    return texto_limpo
  else:
    print(f'Error: {response.status_code}')
    return None
  text = soup.get_text()
  return text

extract_text_from_url('https://dev.to/terrillo/games-design-themselves-from-rts-to-tactical-strategy-game-3fac')
```

```python
from langchain_openai.chat_models.azure import AzureChatOpenAI

client = AzureChatOpenAI(
    azure_endpoint= "AZURE_ENDPOINT",
    api_key= "AZURE_KEY",
    api_version= "",
    deployment_name= "",
    max_retries=0
)

def translate_article(text, lang):
  messages = [
      ("system" , "Você atua como tradutor de textos"),
      ("user" , f"Traduza o {text} para o idioma {lang} e responda o markerdown")
  ]
  response = client.invoke(messages)
  print(response.content)
  return response.content

translate_article("Games design themselves. From RTS to Tactical Strategy game", 'portugues')
```

```python
url = 'https://dev.to/terrillo/games-design-themselves-from-rts-to-tactical-strategy-game-3fac'
text = extract_text_from_url(url)
article = translate_article(text, "portugues")

print(article)
```
